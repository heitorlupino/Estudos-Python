### Forma Básica

```python
precos = "Jan: 25, Fev: 27, Mar: 29"

precos_jan = precos[0:3] # Dois pontos significa até
print(precos_jan)

precos_fev = precos[9:16]
print(precos_fev)
```

### Posição Inicial e Final

```python
precos_mar = precos[-2:]
print(precos_mar)

precos_fev = precos[-11:-9]
print(precos_fev)

precos_fev2 = precos[14:16]
print(precos_fev2)
```

### Posição Inicial e Final com Step

```python
codigo = "1.2.3.4,5,1,2.3.4,7.9"

pedaco_cod = codigo[::4]
print(pedaco_cod)

pedaco_cod2 = codigo[2:10:2]
print(pedaco_cod2)
```
