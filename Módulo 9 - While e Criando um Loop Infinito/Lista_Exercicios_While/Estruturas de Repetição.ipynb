#### 1. Faça um programa que peça uma nota, entre zero e dez. Enquanto o valor corresponder a regra, o usuário vai conseguir escrever quantos números queiser, mas a partir do momento que sair dessa regra, o programa para e é exibido os valores que foram digitados

````python
lista_nota = []
nota = int(input('Digite uma nota entre zero e dez: '))
i = 0

while True:
    if nota < 0 or nota > 10:
        break
    else:
        nota = int(input('Digite uma nota entre zero e dez: '))
        lista_nota.append(nota)
        i += 1

for c in lista_nota:
    print(c)
````

#### 2. Faça um programa que peça uma nota, entre zero e dez. Mostre uma mensagem caso o valor seja inválido e continue pedindo até que o usuário informe um valor válido.

````python
nota = int(input('Digite uma nota entre zero e dez: '))
i = 0

while True:
    if nota < 0 or nota > 10:
        print('Valor inválido. Tente novamente!')
        nota = int(input('Digite uma nota entre zero e dez: '))
        i += 1
    else:
        print(nota)
        break
````

#### 3. Faça um programa que leia um nome de usuário e a sua senha e não aceite a senha igual ao nome do usuário, mostrando uma mensagem de erro e voltando a pedir as informações.

````python
nome_usuario = input('Nome usuário: ')
senha_usuario = input('Digite a sua senha: ')

while True:
    if nome_usuario.replace(' ', '').lower() == senha_usuario.strip().lower():
        print('Senha inválida. tente novamente!')
        nome_usuario = input('Nome usuário: ')
        senha_usuario = input('Digite a sua senha: ')
    else:
        print('O usuário: {}. Senha {}'.format(nome_usuario, senha_usuario))
        break
````

#### 4. Faça um programa que leia e valide as seguintes informações (e para cada uma delas, continue pedindo a informação até o usuário inserir corretamente):
##### Nome: maior que 3 caracteres;
##### Idade: entre 0 e 150;
##### Salário: maior que zero;
##### Sexo: 'f' ou 'm';
##### Estado Civil: 's', 'c', 'v', 'd';

````python
nome = input('Nome: ')
while len(nome) < 4:
        print('Valor inválido. Tente novamente!')
        nome = input('Nome: ')

idade = int(input('Idade: '))
while idade < 0 or idade > 150:
    print('Valor inválido. Tente novamente!')
    idade = int(input('Idade: '))

salario = float(input('Salário: '))
while salario <= 0:
    print('Valor inválido. Tente novamente!')
    salario = float(input('Salário: '))

sexo = input('Sexo (f, m): ')  
while sexo.lower() != 'f' and sexo.lower() != 'm':
    print('valor inválido. Tente novamente!')
    sexo = input('Sexo (f, m): ')

estado_civil = input('Estado civil (s, c, v, d): ')
while estado_civil.lower() != 's' and estado_civil.lower() != 'c' and estado_civil.lower() != 'v' and estado_civil.lower() != 'd':
    print('valor inválido. Tente novamente!')
    estado_civil = input('Estado civil (s, c, v, d): ')
    

print('Nome: {}. Idade: {}. Salário: R${:.2f}. Sexo: {}. Estado civil: {}.'.format(nome, idade, salario, sexo, estado_civil))
````

#### 5. Supondo que a população de um país A seja da ordem de 80000 habitantes com uma taxa anual de crescimento de 3% e que a população de B seja 200000 habitantes com uma taxa de crescimento de 1.5%. Faça um programa que calcule e escreva o número de anos necessários para que a população do país A ultrapasse ou iguale a população do país B, mantidas as taxas de crescimento.

````python
pais_A = 80000
pais_B = 200000
qtde_anos = 0

while pais_A < pais_B:
    pais_A *= 1.03
    pais_B *= 1.015
    qtde_anos += 1
print('Serão necessarios {} anos'.format(qtde_anos))
````
