F-string

```python
faturamento = 1000
custo = 400
lucro = faturamento - custo

#Com format
print('O faturamento foi de {} e o custo de {}'.format(faturamento, lucro))

#Com F-string
print(f"O faturamento foi de {faturamento} e o lucro de {lucro}")
```

Mudança de tipo

```python
faturamento = int(input("Insira o faturamento: "))
custo = int(input("Insira o custo: "))

print(type(faturamento))
print(type(custo))

lucro = faturamento - custo

print(f"O lucro da loja é de {lucro}")
```
