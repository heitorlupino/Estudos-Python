Tipos de Variáveis em Python

Uma variável é um objeto, afinal tudo é um objeto em Python!

Imagine que qualquer coisa que usamos aqui no Python está dentro de uma caixa, dentro dessa caixa possuem propriedades e métodos. Com um texto (string) podemos fazer muitas coisas com ele, como descobrir o tamanho do texto, remover o primeiro caractere do texto, juntar um texto com outro, etc. 

Podemos dizer que temos propriedades / caracteristicas (tamanho do caractere) e métodos diferentes (juntar um texto com outro).

int -> inteiro
flaot -> números com casas decimais (pontos flutuantes)
string -> texto
bool ou boolean -> True or False

Obs: Variáveis em Python não são declaradas explicitamente, mas isso não significa que você deve ficar mudando o tempo todo, pois isso deixa o seu código lento e cair em algum problema.

```python
faturamento = 1000
type(faturamento)
```

```python
faturamento = 1000.00
type(faturamento)
```

```python
faturamento = "1.000"
type(faturamento)
```
